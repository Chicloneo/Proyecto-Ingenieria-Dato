```python
df = pd.read_excel("medicamentos_web_scrapping.xlsx")
```

```python
df_ministerio = pd.read_csv("ministerio_sanidad.csv")
```

```python
for index, row in df.iterrows():
    cn = row['cn']
    resultado = df_ministerio.loc[df_ministerio['Código Nacional'] == cn, 
                                  ["Código Nacional",
                                   "Estado",
                                   "Precio venta al público con IVA", 
                                   "Precio de referencia", 
                                   "Tratamiento de larga duración", 
                                   "Especial control médico"]
                                   ]
```

```python
columnas_anadir = [
    "Código Nacional",
    "Estado",
    "Precio venta al público con IVA",
    "Precio de referencia",
    "Tratamiento de larga duración",
    "Especial control médico",
]
```

```python
df_final = pd.merge(
    df,
    df_ministerio[columnas_anadir],
    left_on="cn", 
    right_on="Código Nacional",
    how="left",  #el merge es de derecha a izquierda. Es decir, mantenemos las columnas de `df``
)
```

```python
df_final.to_excel("df_final.xlsx")
```
