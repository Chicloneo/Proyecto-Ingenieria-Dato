```python
for index, row in df.iterrows():
    url = row["url_html_ficha_tecnica"]
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'html.parser')

    seccion_4_4 = soup.find('h2', id='4.4')
    contenido = seccion_4_4.find_next_sibling()
    texto = contenido.text

    lista_tablas = soup.find_all('table')

    texto = texto.lower()
    coincidencias = re.findall(r"\bgraves?\b", texto)
    
    print(len(texto.split()), len(lista_tablas), len(coincidencias))
```

```python
longitudes_4_4 = []
numero_de_tablas = []
contador_grave = []

for index, row in df.iterrows():
    url = row["url_html_ficha_tecnica"]
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'html.parser')

    seccion_4_4 = soup.find('h2', id='4.4')
    contenido = seccion_4_4.find_next_sibling()
    texto = contenido.text
    longitudes_4_4.append(len(texto.split()))

    lista_tablas = soup.find_all('table')
    numero_de_tablas.append(len(lista_tablas))

    texto = texto.lower()
    coincidencias = re.findall(r"\bgraves?\b", texto)
    contador_grave.append(len(coincidencias))
```

```python
print(longitudes_4_4)
```

```python
print(contador_grave)
```

```python
df = pd.read_excel(ruta_excel)
```

```python
df["Volumen_Informativo"] = longitudes_4_4 
```

Añadimos una columna al dataframe. Como no existe ninguna que se llame Volumen_Informativo, la crea.

```python
df["Complejidad_Desglose"] = numero_de_tablas
```

```python
df["Indicador_Riesgo"] = contador_grave
```

```python
df.to_excel("medicamentos_web_scrapping.xlsx", index=False)
```
