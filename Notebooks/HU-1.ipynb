```python
import pandas as pd
import requests
```

```python
body = [{ 
    "seccion": "4.1", 
    "texto": "migraña", 
    "contiene": 1 
}]

url_post = "https://cima.aemps.es/cima/rest/buscarEnFichaTecnica?pagina=1"
respuesta_post = requests.post(url=url_post, json=body).json()
resultados = respuesta_post["resultados"]
#resultados
```

```python
url_get = "https://cima.aemps.es/cima/rest/medicamento"
respuesta_get = requests.get(url=url_get,params={'nregistro':78686}).json()
```

```python
caracteristicas = [
    "nregistro",
    "nombre",
    "pactivos",
    "labtitular",
    "labcomercializador",
    "cn",
    "dosis",
    "forma_farmaceutica_simplificada",
    "estado_aut",
    "estado_rev",
    "vias_administracion",
    "comercializado",
    "requiere_receta",
    "generico",
    "afecta_conduccion",
    "triangulo_negro",
    "medicamento_huerfano",
    "biosimilar",
    "url_html_ficha_tecnica",
    "url_foto_materiales",
    "num_registros_atc",
    "num_principios_activos",
    "num_excipientes"
]
```

```python
caracteristicas_problems = {
    "cn": respuesta_get["presentaciones"][0]["cn"],
    "forma_farmaceutica_simplificada": respuesta_get["formaFarmaceuticaSimplificada"]["nombre"],
    "estado_aut": respuesta_get["estado"]["aut"],
    "vias_administracion": respuesta_get["viasAdministracion"][0]["nombre"],
    "comercializado": int(respuesta_get["comerc"]),
    "requiere_receta": int(respuesta_get["receta"]),
    "afecta_conduccion": int(respuesta_get["conduc"]),
    "triangulo_negro": int(respuesta_get["triangulo"]),
    "medicamento_huerfano": int(respuesta_get["huerfano"]),
    "url_html_ficha_tecnica": respuesta_get["docs"][0]["urlHtml"],
    "url_foto_materiales": respuesta_get["fotos"][0]["url"],
    "num_registros_atc": len(respuesta_get["atcs"]),
    "num_principios_activos": len(respuesta_get["principiosActivos"]),
    }
```

```python
for caracteristica in caracteristicas:
    try:
        respuesta_get[caracteristica]
    except:
        if caracteristica == "estado_rev":
            try:
                respuesta_get["estado"]["rev"]
            except:
                None
        elif caracteristica == "num_excipientes":
                    try:
                        len(respuesta_get["excipientes"])
                    except:
                        None
        else:
            caracteristicas_problems[caracteristica]
```

```python
lista_carcteristicas = []

for caracteristica in caracteristicas:
    try:
        lista_carcteristicas.append(respuesta_get[caracteristica])
    except:
        try:
            lista_carcteristicas.append(caracteristicas_problems[caracteristica])
        except:
            lista_carcteristicas.append(None)
```
