```python
import pandas as pd
import requests
import json
```

```python
body = [{ 
    "seccion": "4.1", 
    "texto": "migraña", 
    "contiene": 1 
}]

url_post = "https://cima.aemps.es/cima/rest/buscarEnFichaTecnica?pagina=1"
respuesta_post = requests.post(url=url_post, json=body).json()
resultados = respuesta_post["resultados"]
#resultados
```

```python
nregistros = []
for medicamento in resultados:
    nregistros.append(medicamento["nregistro"])
#nregistros
```

```python
url_get = "https://cima.aemps.es/cima/rest/medicamento"
respuesta_get = requests.get(url=url_get,params={'nregistro':78686})
json_medicamentos = respuesta_get.json()
#json_medicamentos
```

```python
medicamentos = []
for registro in nregistros:
    medicamentos.append(requests.get(url=url_get,params={'nregistro':registro}).json())
#medicamentos
```

```python
df_lista = []

for medicamento in medicamentos:
    lista_carcteristicas = []

    caracteristicas_calls = {
        "cn": medicamento.get("presentaciones")[0].get("cn"),
        "forma_farmaceutica_simplificada": medicamento.get("formaFarmaceuticaSimplificada").get("nombre"),
        "estado_aut": medicamento.get("estado").get("aut"),
        "estado_rev": medicamento.get("estado").get("rev"),
        "vias_administracion": medicamento.get("viasAdministracion")[0].get("nombre"),
        "comercializado": int(medicamento.get("comerc")),
        "requiere_receta": int(medicamento.get("receta")),
        "afecta_conduccion": int(medicamento.get("conduc")),
        "triangulo_negro": int(medicamento.get("triangulo")),
        "medicamento_huerfano": int(medicamento.get("huerfano")),
        "url_html_ficha_tecnica": medicamento.get("docs")[0].get("urlHtml"),
        "url_foto_materiales": medicamento.get("fotos")[0].get("url") if "fotos" in medicamento else None,
        "num_registros_atc": len(medicamento.get("atcs")),
        "num_principios_activos": len(medicamento.get("principiosActivos")),
        "num_excipientes": len(medicamento.get("excipientes")) if not pd.isna(medicamento.get("excipientes")) else pd.NA,
        "nregistro": medicamento.get("nregistro"),
        "nombre": medicamento.get("nombre"),
        "pactivos": medicamento.get("pactivos"),
        "labtitular": medicamento.get("labtitular"),
        "labcomercializador": medicamento.get("labcomercializador"),
        "dosis": medicamento.get("dosis"),
        "generico": medicamento.get("generico"),
        "biosimilar": medicamento.get("biosimilar"),
    }
    
    medicamento_dict={}
    for caracteristica in CARACTERISTICAS:
        medicamento_dict[caracteristica] = caracteristicas_calls[caracteristica]
        #lista_carcteristicas.append(caracteristicas_calls[caracteristica])

    df_lista.append(medicamento_dict)

df_lista
```

```python
df = pd.DataFrame(df_lista)
df.head()
df.to_csv("../src/data/HU-1.csv", index=False)
```
