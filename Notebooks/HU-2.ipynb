```python
ruta_csv = "../src/data/HU-1.csv"
df = pd.read_csv(ruta_csv)
df["url_html_ficha_tecnica"].head()
```

```python
volumenes = []
complejidades = []
indicadores = []

for index, row in df.iterrows():
    link = row["url_html_ficha_tecnica"]
    response = requests.get(link).text
    #print(response)
    soup = BeautifulSoup(response, 'html.parser')

    seccion_4_4 = soup.find(id='4.4')
    contenido_4_4 = seccion_4_4.find_next()
    palabras_4_4 = contenido_4_4.get_text(strip = True).split()
    num_palabras = len(palabras_4_4)
    volumenes.append(num_palabras)

    tablas = soup.find_all('table')
    num_tablas = len(tablas)
    complejidades.append(num_tablas)

    texto_medicamento = soup.get_text(" ", strip=True).lower()
    lista_graves = re.findall(r'\bgraves?\b', texto_medicamento)
    num_graves = len(lista_graves)
    indicadores.append(num_graves)

df["Volumen_Informativo"] = volumenes
df["Complejidad"] = complejidades
df["Indicador_Riesgo"] = indicadores

#df.head()
```
