```python
import pandas as pd
```

```python
ruta = "../src/data/HU-2.csv"
df_migraña = pd.read_csv(ruta)
```

```python
ruta = "../src/data/Nomenclator.csv"
df_total = pd.read_csv(ruta)
```

```python
df_final = pd.merge(df_migraña, df_total[["Código Nacional","Estado","Precio venta al público con IVA","Precio de referencia", \
        "Tratamiento de larga duración", "Especial control médico"]], left_on = 'cn', right_on='Código Nacional', how="left")
df_final.head()
```

```python
df_final = pd.merge(left=df_migraña, right=df_total[["Código Nacional","Estado","Precio venta al público con IVA","Precio de referencia", \
        "Tratamiento de larga duración", "Especial control médico"]], left_on='cn', right_on='Código Nacional', how='left')
df_final.head()
```

```python
ruta_output = "../src/data/HU-3.csv"
df_final.to_csv(ruta_output, index=False)
```
